****
# iloc[]

-  TO print specific rows and columns we use **data_name.iloc[0:3,0:6]**
    -  (0:3 means form **0th index** to **3-1 row** and 0:6 means from **0th to 6-1 columns**)
    -  You can use **'df.iloc[:3]'** this if you want all columns and only specific rows.
    -  You can use **'df.iloc[:,:3]'** this if you want all rows and only specific columns.
    -  **df.iloc[:3,0:6:2]** Using this you can skip 1 columns and fetch and agin skip 1 column

```python
df.iloc[0:3,0:6]  
```

```python
df.iloc[:2,:3]
```

```python
df.iloc[:3]
```

```python
df.iloc[:4,0:6:2]
```

```python
df.iloc[:7:3,0:6:2]
```

****
# loc[]

- How to use loc[]
    - Use to fetch only specific rows and columns
    - **df.loc[:3,['columns name to fetch in single code']]** This is the command
    - **Note**: In this function you dont't need to give index+1 value
        - for example: df.loc[:3], this will print values from 0th index to 3th index unlike other slicing methods.

```python
df.loc[:3,['Petrol_tax','Paved_Highways']]
```

```python
df.loc[11:15,['Petrol_tax','Petrol_Consumption']]
```
